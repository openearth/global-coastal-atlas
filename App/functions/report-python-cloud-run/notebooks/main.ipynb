### Packages

```python
# Packages
from report.report import generate_report_content, create_report_pdf, create_report_html
from shapely.geometry import box
from resilientplotterclass import rpc
import geopandas as gpd
import warnings
```

### Specify polygon

```python
# Specify polygon
polygon = box(5.100403, 53.324312, 5.571442, 53.48396)
gdf_polygon = gpd.GeoDataFrame(geometry=[polygon], crs="EPSG:4326")

# Initialise map
m = rpc.draw_map(gdf=gdf_polygon)

# Show map
m
```

```python
# Get polygon from map
gdf_polygon = m.get_geometries()
polygon = gdf_polygon.geometry.iloc[0]
```

### Generate report

```python
# Generate report content
# report_content = generate_report_content(polygon)
```

```python
# Generate report html
with warnings.catch_warnings():
    warnings.filterwarnings("ignore", module="pystac_client")
    report_html = create_report_html(polygon)

# Save report html to file
with open("report.html", "w", encoding="utf-8") as f:
    f.write(report_html)
```

### Retrieval-Augmented Generation (RAG)

```python
from dllmforge.rag_search_and_response import LLMResponder
from dllmforge.langchain_api import LangchainAPI
from report.utils.rag import set_up_RAG_AZURE

# Initialise large language model and retriever
retriever = set_up_RAG_AZURE("minikennisbank", index_exists=True, index_name="minikennisbank-index-antonio")
responder = LLMResponder(LangchainAPI(model_provider="azure-openai", temperature=0.4).llm)

# Ask a question
question = r"Have we done a study in the Gennaker offshore wind farm area and what are the main findings?"
retrieved_contexts = retriever.invoke(question)
answer = responder.generate(question, retrieved_contexts)

# Show answer
print(answer)
```
